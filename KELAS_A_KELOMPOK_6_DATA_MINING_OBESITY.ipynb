Multikolinearitas (VIF)

---
Mengecek apakah ada multikolinearitas antar variabel independen

```python
# Memilih kolom numerik spesifik yang akan dianalisis
kolom_vif = ["Age", "Height", "Weight", "FCVC", "NCP", "CH2O", "FAF", "TUE"]
kolom_vif_exist = [col for col in kolom_vif if col in df.columns]

# Menyiapkan data numerik
X_vif_data = df[kolom_vif_exist].apply(pd.to_numeric, errors="coerce")
X_vif_data = X_vif_data.replace([np.inf, -np.inf], np.nan).dropna()

# Menambahkan konstanta (intercept) untuk perhitungan VIF
X_vif_with_const = add_constant(X_vif_data)

# Menghitung VIF
vif_df = pd.DataFrame()
vif_df["Feature"] = X_vif_with_const.columns
vif_df["VIF"] = [
    variance_inflation_factor(X_vif_with_const.values.astype(float), i)
    for i in range(len(X_vif_with_const.columns))
]

# Menghilangkan baris konstanta untuk tampilan
vif_df_final = vif_df[vif_df["Feature"] != "const"]

print("Hasil Perhitungan Variance Inflation Factor (VIF):")
display(vif_df_final.sort_values(by="VIF", ascending=False).style.background_gradient(cmap='RdPu_r'))
```

Berdasarkan hasil perhitungan *Variance Inflation Factor* (VIF), seluruh variabel numerik memiliki nilai VIF antara 1,064 hingga 1,567, yang berada di bawah batas 10. Hal ini menunjukkan bahwa **tidak terdapat indikasi multikolinearitas tinggi** antarvariabel numerik yang dianalisis, sehingga seluruh variabel tersebut dapat dipertahankan untuk tahap pemodelan.

Deteksi Outlier [Univariat dengan IQR Method]

---


```python
# DETEKSI OUTLIER PADA VARIABEL KONTINU DI DATA LATIH (X_train)
# Memilih hanya variabel yang bersifat kontinu
kolom_kontinu = ['Age', 'Height', 'Weight']

print("=== DETEKSI OUTLIER PADA VARIABEL KONTINU (IQR METHOD) ===\n")

for col in kolom_kontinu:
    if col in X_train.columns:
        data_col = X_train[col]

        # Menghitung Kuartil 1 (Q1), Kuartil 3 (Q3), dan Interquartile Range (IQR)
        Q1 = data_col.quantile(0.25)
        Q3 = data_col.quantile(0.75)
        IQR = Q3 - Q1

        # Menentukan Batas Bawah dan Batas Atas
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        # Filter data yang berada di luar batas
        outliers = X_train[(data_col < lower_bound) | (data_col > upper_bound)]

        print(f"Variabel: {col}")
        print(f"  -> Q1: {Q1:.4f} | Q3: {Q3:.4f} | IQR: {IQR:.4f}")
        print(f"  -> Batas Bawah (Lower Bound): {lower_bound:.4f}")
        print(f"  -> Batas Atas (Upper Bound): {upper_bound:.4f}")
        print(f"  -> Jumlah Outlier: {len(outliers)} baris")

        if len(outliers) > 0:
            print("  -> Daftar Data Outlier:")
            display(outliers[[col]])
        else:
            print("  -> Tidak ditemukan data outlier pada variabel ini.")
        print("-" * 60 + "\n")
```

data yang dicek outlier hanya variabel kontinu saja


Hasil pemeriksaan outlier dengan metode IQR pada data latih (1.667 observasi) menunjukkan 124 outlier pada `Age` (di atas batas atas 35 tahun), 1 outlier pada `Height` (di atas 1,9787 meter), dan tidak ada outlier pada `Weight`. Nilai-nilai tersebut masih mungkin secara biologis sehingga tidak dihapus. Pada model utama, data dipakai apa adanya (dengan outlier), sedangkan penanganan outlier dengan capping hanya dilakukan pada eksperimen tambahan Naive Bayes.

Penanganan Outlier untuk Algoritma Naive Bayes

```python
# 1. Membuat salinan data training khusus untuk penanganan outlier Naive Bayes
X_train_nb_clean = X_train.copy()
y_train_nb_clean = y_train.copy()

# Variabel kontinu yang akan ditangani
kolom_kontinu = ['Age', 'Height', 'Weight']

print("=== PROSES PENANGANAN OUTLIER PADA DATA TRAIN (CAPPING METHOD) ===\n")

for col in kolom_kontinu:
    if col in X_train_nb_clean.columns:
        # Menghitung kuartil dan batas IQR dari data training asli
        Q1 = X_train[col].quantile(0.25)
        Q3 = X_train[col].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        # Melakukan capping (winsorizing) hanya pada salinan data training
        X_train_nb_clean[col] = X_train_nb_clean[col].clip(lower=lower_bound, upper=upper_bound)

        print(f"Variabel '{col}':")
        print(f"  -> Batas bawah: {lower_bound:.4f} | Batas atas: {upper_bound:.4f}")
        print(f"  -> Nilai pada 'X_train_nb_clean' telah berhasil di-cap.\n")

print("Selesai! Salinan data bersih disimpan dalam variabel 'X_train_nb_clean'.")
```

Penanganan outlier dilakukan dengan metode ***capping* (*winsorizing*)**: nilai di bawah batas bawah diganti dengan batas bawah, dan nilai di atas batas atas diganti dengan batas atas. Dengan cara ini jumlah observasi data latih tetap **1.667 baris**, tidak ada data yang hilang.

Beberapa hal penting dari proses ini:

- Batas bawah dan atas dihitung **hanya dari data latih**, dan capping hanya diterapkan pada salinan `X_train_nb_clean`. Data uji tidak diubah sehingga evaluasi tidak bocor dan tetap mencerminkan data apa adanya.
- *Capping* dipilih dibanding penghapusan baris karena penghapusan dapat mengurangi data pada kelas tertentu (misalnya kelas obesitas berat yang wajar memiliki berat badan ekstrem) dan mengganggu keseimbangan kelas.
- Penanganan ini difokuskan untuk Naive Bayes karena model ini mengasumsikan distribusi Gaussian, sehingga nilai ekstrem dapat menggeser rata-rata dan varians tiap kelas.

Feature Engineering

Feature Engineering mengubah data mentah menjadi data yang siap diproses model. Encoding dan scaling dilakukan pada dua versi data latih: data dengan outlier (X_train, dipakai untuk perbandingan utama) dan data hasil capping (X_train_nb_clean, dipakai hanya untuk eksperimen tambahan Naive Bayes).

Encoding Kategorikal: variabel biner (`Gender`, `family_history_with_overweight`, `FAVC`, `SMOKE`, `SCC`) diubah menjadi 0/1 dengan Label Encoder; variabel ordinal (`CAEC` dan `CALC`) dikodekan berurutan (no=0 < Sometimes=1 < Frequently=2 < Always=3); variabel nominal `MTRANS` diubah dengan One-Hot Encoding; target dikodekan 0 sampai 6 berurutan menurut tingkat BMI.

Scaling Numerik: StandardScaler diterapkan pada `Age`, `Height`, `Weight`, `FCVC`, `NCP`, `CH2O`, `FAF`, dan `TUE`. Scaler dipelajari hanya dari data latih masing-masing versi, lalu dipakai untuk mentransformasi data uji. Karena itu data uji memiliki dua versi scaling mengikuti scaler dari data latih yang dipakai model.
Scaling diterapkan pada semua model agar praproses seragam. Skala fitur tidak memengaruhi Decision Tree, tetapi penting untuk SVM.

```python
# 1. Salinan Data untuk Proses Encoding
X_train_enc = X_train.copy()
X_train_nb_clean_enc = X_train_nb_clean.copy()
X_test_enc = X_test.copy()

# 2. Encoding untuk Variabel Biner
bin_cols = ['Gender', 'family_history_with_overweight', 'FAVC', 'SMOKE', 'SCC']
map_biner = {'yes': 1, 'no': 0, 'Male': 1, 'Female': 0}

for col in bin_cols:
    X_train_enc[col] = X_train_enc[col].map(map_biner).astype(int)
    X_train_nb_clean_enc[col] = X_train_nb_clean_enc[col].map(map_biner).astype(int)
    X_test_enc[col] = X_test_enc[col].map(map_biner).astype(int)

# 3. Ordinal Encoding untuk CAEC & CALC
map_ordinal = {'no': 0, 'Sometimes': 1, 'Frequently': 2, 'Always': 3}
ord_cols = ['CAEC', 'CALC']

for col in ord_cols:
    X_train_enc[col] = X_train_enc[col].map(map_ordinal)
    X_train_nb_clean_enc[col] = X_train_nb_clean_enc[col].map(map_ordinal)
    X_test_enc[col] = X_test_enc[col].map(map_ordinal)
    assert X_train_enc[col].notna().all() and X_test_enc[col].notna().all(), f"Ada kategori tak terpetakan di {col}"

# 4. One-Hot Encoding untuk MTRANS
# Skenario A: Data dengan Outlier
ohe_sd = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
mtrans_train_sd_ohe = ohe_sd.fit_transform(X_train_enc[['MTRANS']])
ohe_cols_sd = [f"MTRANS_{cat}" for cat in ohe_sd.categories_[0]]
df_mtrans_train_sd = pd.DataFrame(mtrans_train_sd_ohe, columns=ohe_cols_sd, index=X_train_enc.index)

# Transform ke test set untuk skenario dengan outlier
mtrans_test_sd_ohe = ohe_sd.transform(X_test_enc[['MTRANS']])
df_mtrans_test_sd = pd.DataFrame(mtrans_test_sd_ohe, columns=ohe_cols_sd, index=X_test_enc.index)

# Skenario B: Data tanpa Outlier
ohe_cl = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
mtrans_train_cl_ohe = ohe_cl.fit_transform(X_train_nb_clean_enc[['MTRANS']])
ohe_cols_cl = [f"MTRANS_{cat}" for cat in ohe_cl.categories_[0]]
df_mtrans_train_cl = pd.DataFrame(mtrans_train_cl_ohe, columns=ohe_cols_cl, index=X_train_nb_clean_enc.index)

# Transform ke test set untuk skenario tanpa outlier
mtrans_test_cl_ohe = ohe_cl.transform(X_test_enc[['MTRANS']])
df_mtrans_test_cl = pd.DataFrame(mtrans_test_cl_ohe, columns=ohe_cols_cl, index=X_test_enc.index)

# Menggabungkan hasil OHE ke DataFrame masing-masing
X_train_enc = pd.concat([X_train_enc.drop(columns=['MTRANS']), df_mtrans_train_sd], axis=1)
X_train_nb_clean_enc = pd.concat([X_train_nb_clean_enc.drop(columns=['MTRANS']), df_mtrans_train_cl], axis=1)

# Kita siapkan dua versi X_test untuk masing-masing skenario model
X_test_sd_enc = pd.concat([X_test_enc.drop(columns=['MTRANS']), df_mtrans_test_sd], axis=1)
X_test_cl_enc = pd.concat([X_test_enc.drop(columns=['MTRANS']), df_mtrans_test_cl], axis=1)

# 5. Encoding Target NObeyesdad (0-6)
map_target = {
    'Insufficient_Weight': 0,
    'Normal_Weight': 1,
    'Overweight_Level_I': 2,
    'Overweight_Level_II': 3,
    'Obesity_Type_I': 4,
    'Obesity_Type_II': 5,
    'Obesity_Type_III': 6
}

y_train_sd_enc = y_train.map(map_target).astype(int)
y_train_cl_enc = y_train_nb_clean.map(map_target).astype(int)
y_test_enc = y_test.map(map_target).astype(int)

print("Proses encoding selesai!")
print("Dimensi X_train (dengan outlier) :", X_train_enc.shape)
print("Dimensi X_train (tanpa outlier)  :", X_train_nb_clean_enc.shape)
print("Dimensi X_test (versi outlier)   :", X_test_sd_enc.shape)
print("Dimensi X_test (versi bersih)    :", X_test_cl_enc.shape)

print("\n--- Menampilkan Dataframe Hasil Encoding ---")
print("\n1. DataFrame X_train_enc (Dengan Outlier) [5 Baris Pertama]:")
display(X_train_enc.head())

print("\n2. DataFrame X_train_nb_clean_enc (Tanpa Outlier) [5 Baris Pertama]:")
display(X_train_nb_clean_enc.head())

print("\n3. DataFrame X_test_sd_enc (Versi Outlier) [5 Baris Pertama]:")
display(X_test_sd_enc.head())

print("\n4. DataFrame X_test_cl_enc (Versi Bersih) [5 Baris Pertama]:")
display(X_test_cl_enc.head())
```

*Encoding* mengubah seluruh variabel non-numerik menjadi angka: variabel biner (`Gender`, `family_history_with_overweight`, `FAVC`, `SMOKE`, `SCC`) menjadi 0/1, variabel ordinal (`CAEC`, `CALC`) dikodekan berurutan (`no` = 0 hingga `Always` = 3), `MTRANS` diubah dengan *One-Hot Encoding* menjadi 5 kolom, dan target `NObeyesdad` dikodekan 0 sampai 6 sesuai tingkat BMI.

Hasilnya, data latih (dengan maupun tanpa outlier) berdimensi **1.667 x 20** dan data uji **417 x 20**. Jumlah kolom bertambah dari 16 menjadi 20 karena `MTRANS` diperluas menjadi 5 kolom, sedangkan jumlah baris tidak berubah. Seluruh kolom kini bertipe numerik sehingga siap di-*scaling*.

Scaling (StandardScaler), dipelajari hanya dari data latih

```python
# Kolom numerik yang akan di-scale
numeric_cols_to_scale = ['Age', 'Height', 'Weight', 'FCVC', 'NCP', 'CH2O', 'FAF', 'TUE']

# 1. Skenario A: Data dengan Outlier
scaler_sd = StandardScaler()
X_train_sd = X_train_enc.copy()
X_test_sd = X_test_sd_enc.copy()

X_train_sd[numeric_cols_to_scale] = scaler_sd.fit_transform(X_train_enc[numeric_cols_to_scale])
X_test_sd[numeric_cols_to_scale] = scaler_sd.transform(X_test_sd_enc[numeric_cols_to_scale])

# 2. Skenario B: Data tanpa Outlier
scaler_cl = StandardScaler()
X_train_cl = X_train_nb_clean_enc.copy()
X_test_cl = X_test_cl_enc.copy()

X_train_cl[numeric_cols_to_scale] = scaler_cl.fit_transform(X_train_nb_clean_enc[numeric_cols_to_scale])
X_test_cl[numeric_cols_to_scale] = scaler_cl.transform(X_test_cl_enc[numeric_cols_to_scale])

# Target encoding yang sesuai
y_train_sd = y_train_sd_enc
y_train_cl = y_train_cl_enc
y_test_sd = y_test_enc
y_test_cl = y_test_enc

print("Proses scaling selesai!")
print("Dimensi data latih dengan outlier (X_train_sd) :", X_train_sd.shape)
print("Dimensi data latih tanpa outlier (X_train_cl)  :", X_train_cl.shape)
print("Dimensi data uji (X_test_sd)                 :", X_test_sd.shape)
print("Dimensi data uji (X_test_cl)                 :", X_test_cl.shape)
```

*Scaling* dilakukan dengan `StandardScaler` pada delapan variabel numerik (`Age`, `Height`, `Weight`, `FCVC`, `NCP`, `CH2O`, `FAF`, `TUE`) dalam dua skenario: **A** (data dengan outlier) dan **B** (data tanpa outlier hasil *capping*). *Scaler* dipelajari hanya dari data latih (`fit_transform`), sedangkan data uji hanya di-*transform* sehingga tidak ada kebocoran data.

Dimensi data tidak berubah: data latih skenario A dan B masing-masing **1.667 x 20**, dan data uji masing-masing **417 x 20**.

Model Pertama: Naive Bayes

Model pertama adalah Gaussian Naive Bayes (GaussianNB), varian Naive Bayes untuk fitur numerik yang mengasumsikan setiap fitur berdistribusi normal pada tiap kelas dan saling bebas. Model dijalankan dengan parameter default (var_smoothing = 1e-9).

```python
# 1. Inisialisasi model Naive Bayes dengan parameter default
model_nb_default = GaussianNB()

# 2. Melatih model menggunakan data latih ter-scale dengan outlier
model_nb_default.fit(X_train_sd, y_train_sd)

# 3. Prediksi pada data uji (test set)
y_test_pred = model_nb_default.predict(X_test_sd)

# 4. Menghitung Metrik Evaluasi Akhir
accuracy = accuracy_score(y_test_sd, y_test_pred)
precision_macro = precision_score(y_test_sd, y_test_pred, average='macro', zero_division=0)
recall_macro = recall_score(y_test_sd, y_test_pred, average='macro', zero_division=0)
f1_macro = f1_score(y_test_sd, y_test_pred, average='macro', zero_division=0)

# 5. Menampilkan Hasil Evaluasi Akhir
print("=== EVALUASI AKHIR NAIVE BAYES (DEFAULT) PADA DATA UJI ===")
print(f"Accuracy        : {accuracy:.4f}")
print(f"Precision Macro : {precision_macro:.4f}")
print(f"Recall Macro    : {recall_macro:.4f}")
print(f"F1-Score Macro  : {f1_macro:.4f}\n")

print("Laporan Klasifikasi Lengkap (Classification Report) Data Uji:")
print(classification_report(y_test_sd, y_test_pred, target_names=urutan_kelas, zero_division=0))

# 6. Visualisasi Confusion Matrix
plt.figure(figsize=(8, 6))
sns.heatmap(confusion_matrix(y_test_sd, y_test_pred), annot=True, fmt='d', cmap='YlGnBu',
            xticklabels=urutan_kelas, yticklabels=urutan_kelas)
plt.title("Confusion Matrix - Naive Bayes Default (Data Uji)")
plt.ylabel("Aktual (Data Uji)")
plt.xlabel("Prediksi (Data Uji)")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

Berdasarkan *classification report* dan *confusion matrix*, model paling baik mengklasifikasikan **Obesity_Type_III** dengan F1-score **0,98**, yaitu 64 dari 65 observasi terklasifikasi dengan benar. Sebaliknya, model masih kesulitan membedakan kelas **Normal_Weight**, **Overweight_Level_I**, dan **Overweight_Level_II**, yang memiliki F1-score masing-masing **0,15**, **0,14**, dan **0,12**.

Secara keseluruhan, performa Naive Bayes Default dengan outlier belum optimal dan dapat dijadikan pembanding untuk mengevaluasi hasil model setelah penanganan outlier.

```python
# 1. Evaluasi Model Dengan Height & Weight (Menggunakan data latih dan uji ter-scaling penuh)
# Naive Bayes
model_nb_with = GaussianNB()
model_nb_with.fit(X_train_sd, y_train_sd)
y_pred_nb_with = model_nb_with.predict(X_test_sd)

# Decision Tree
model_dt_with = DecisionTreeClassifier(random_state=42)
model_dt_with.fit(X_train_sd, y_train_sd)
y_pred_dt_with = model_dt_with.predict(X_test_sd)

# SVM
model_svm_with = SVC(random_state=42)
model_svm_with.fit(X_train_sd, y_train_sd)
y_pred_svm_with = model_svm_with.predict(X_test_sd)

# 2. Evaluasi Model Tanpa Height & Weight
# Naive Bayes
model_nb_without = GaussianNB()
model_nb_without.fit(X_train_no_hw, y_train_sd)
y_pred_nb_without = model_nb_without.predict(X_test_no_hw)

# Decision Tree
model_dt_without = DecisionTreeClassifier(random_state=42)
model_dt_without.fit(X_train_no_hw, y_train_sd)
y_pred_dt_without = model_dt_without.predict(X_test_no_hw)

# SVM
model_svm_without = SVC(random_state=42)
model_svm_without.fit(X_train_no_hw, y_train_sd)
y_pred_svm_without = model_svm_without.predict(X_test_no_hw)

# 3. Membuat Tabel Perbandingan
list_perbandingan = []

def tambah_metrik(nama_model, skenario, y_true, y_pred):
    list_perbandingan.append({
        'Model': nama_model,
        'Skenario': skenario,
        'Accuracy': accuracy_score(y_true, y_pred),
        'Precision (Macro)': precision_score(y_true, y_pred, average='macro', zero_division=0),
        'Recall (Macro)': recall_score(y_true, y_pred, average='macro', zero_division=0),
        'F1-Score (Macro)': f1_score(y_true, y_pred, average='macro', zero_division=0)
    })

tambah_metrik('Naive Bayes', 'Dengan Height & Weight', y_test_sd, y_pred_nb_with)
tambah_metrik('Naive Bayes', 'Tanpa Height & Weight', y_test_sd, y_pred_nb_without)
tambah_metrik('Decision Tree', 'Dengan Height & Weight', y_test_sd, y_pred_dt_with)
tambah_metrik('Decision Tree', 'Tanpa Height & Weight', y_test_sd, y_pred_dt_without)
tambah_metrik('SVM', 'Dengan Height & Weight', y_test_sd, y_pred_svm_with)
tambah_metrik('SVM', 'Tanpa Height & Weight', y_test_sd, y_pred_svm_without)

df_compare = pd.DataFrame(list_perbandingan)

print("=== TABEL PERBANDINGAN PERFORMA MODEL (DENGAN VS TANPA HEIGHT & WEIGHT) ===")
display(df_compare.style.format({
    'Accuracy': '{:.2%}',
    'Precision (Macro)': '{:.2%}',
    'Recall (Macro)': '{:.2%}',
    'F1-Score (Macro)': '{:.2%}'
}).background_gradient(cmap='Blues', subset=['Accuracy', 'F1-Score (Macro)']))

# 4. Visualisasi Grafik Perbandingan Akurasi
plt.figure(figsize=(12, 6))
sns.barplot(data=df_compare, x='Model', y='Accuracy', hue='Skenario', palette='Set2')
plt.title('Perbandingan Akurasi Model dengan vs tanpa Height & Weight', fontsize=14, fontweight='bold', pad=15)
plt.ylabel('Akurasi')
plt.ylim(0, 1.05)
for p in plt.gca().patches:
    height = p.get_height()
    if height > 0:
        plt.gca().annotate(f'{height:.2%}', (p.get_x() + p.get_width() / 2., height + 0.02),
                            ha='center', va='center', fontsize=10, xytext=(0, 5), textcoords='offset points')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

Penghapusan `Height` dan `Weight` menurunkan akurasi ketiga model: Naive Bayes dari 52,04% menjadi 42,69% (turun 9,35 poin), Decision Tree dari 95,44% menjadi 71,46% (turun 23,98 poin), dan SVM dari 90,65% menjadi 73,62% (turun 17,03 poin). Penurunan terbesar pada Decision Tree dan SVM terjadi karena kedua model tersebut sebelumnya sangat bergantung pada `Weight` dan `Height`, yang merupakan komponen penyusun BMI sebagai dasar penentuan label (target leakage). Pada skenario tanpa H&W, model hanya memakai kebiasaan makan, aktivitas fisik, dan karakteristik demografis sehingga angka 71–74% lebih mencerminkan kemampuan model membaca pola kebiasaan hidup. SVM (73,62%) sedikit lebih tinggi dari Decision Tree (71,46%) dengan selisih sekitar 9 observasi dari 417 data uji, sehingga keunggulannya tipis dan belum tentu konsisten pada pembagian data yang lain.

# **Keterbatasan**

Keterbatasan penelitian:
1. **Sekitar 77% data merupakan data sintetis hasil SMOTE**. Baris sintetis dibentuk dari interpolasi data asli sehingga baris yang mirip dapat berada di data latih maupun data uji. Akibatnya, akurasi yang diperoleh bisa lebih optimis dibanding kinerja pada data nyata.
2. **Label target dibentuk dari BMI (berat dan tinggi badan)**, sehingga skenario dengan Height dan Weight mengandung target leakage.
3. **Seluruh model memakai parameter default dan evaluasi berasal dari satu kali pembagian data** (417 data uji), sehingga selisih kinerja yang kecil antarmodel tidak dapat dianggap pasti.

# **Kesimpulan**

Berdasarkan hasil analisis klasifikasi tingkat obesitas menggunakan Gaussian Naive Bayes, Decision Tree, dan Support Vector Machine (SVM) dengan parameter default dan praproses yang sama, diperoleh kesimpulan sebagai berikut.

1. Setelah pembersihan data diperoleh 2.084 observasi yang dibagi menjadi 1.667 data latih (80%) dan 417 data uji (20%) secara stratifikasi. Variabel `FCVC`, `NCP`, `CH2O`, `FAF`, `TUE`, dan `Age` dibulatkan sesuai karakteristik datanya, sedangkan `Height` dan `Weight` dibiarkan kontinu. Scaling dan capping dihitung hanya dari data latih. Outlier ditemukan pada `Age` (124 data) dan `Height` (1 data), namun masih wajar secara biologis sehingga data utama dipakai tanpa penanganan outlier.
2. Pada data uji dengan seluruh fitur, Decision Tree memperoleh kinerja tertinggi (akurasi 95,44%, F1 Macro 95,29%), diikuti SVM (akurasi 90,65%, F1 Macro 90,19%) dan Naive Bayes (akurasi 52,04%, F1 Macro 43,67%). Hasil ini sebagian besar dipengaruhi oleh `Height` dan `Weight` yang menyusun BMI sebagai dasar penentuan label (target leakage), sehingga tidak sepenuhnya menunjukkan kemampuan model mempelajari kebiasaan makan dan kondisi fisik.
Pada eksperimen tambahan, penanganan outlier dengan capping tidak memberikan perbedaan yang bermakna pada Naive Bayes (akurasi 52,04% menjadi 52,52%, selisih sekitar 2 observasi).
3. Decision Tree mencapai akurasi latih 100% yang menunjukkan pohon menghafal data latih. Namun dengan seluruh fitur selisih latih dan ujinya (4,56 poin) sebanding dengan SVM (4,91 poin). *Overfitting* Decision Tree baru tampak jelas setelah `Height` dan `Weight` dihapus (selisih 27,10 poin). Naive Bayes memiliki selisih terkecil karena underfitting, yaitu kinerjanya rendah pada data latih maupun uji.
4. Setelah `Height` dan `Weight` dihapus, akurasi turun pada seluruh model: Naive Bayes 42,69%, Decision Tree 71,46%, dan SVM 73,62%. SVM sedikit lebih unggul dari Decision Tree (selisih sekitar 9 observasi) dengan generalisasi yang lebih terjaga. Hasil ini lebih mencerminkan kemampuan model memprediksi tingkat obesitas dari kebiasaan hidup.
5. Secara umum, SVM dan Decision Tree jauh lebih baik dibanding Naive Bayes pada data ini karena Naive Bayes mengasumsikan fitur saling bebas dan berdistribusi Gaussian. Hasil perlu ditafsirkan dengan hati-hati karena sekitar 77% data bersifat sintetis (SMOTE) sehingga akurasi dapat terlalu optimis, dan perbandingan dilakukan pada satu kali pembagian data dengan parameter default.
